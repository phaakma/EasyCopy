# Test EasyCopy  

Test using real data.  

```python
import arcpy
from easycopy import EasyCopy
```

```python
source = arcpy.MakeFeatureLayer_management(
    r"C:\projects\playground\playground.gdb\easycopy_source")
target = arcpy.MakeFeatureLayer_management(
    r"C:\projects\playground\playground.gdb\easycopy_target")

print(source)
print(target)
```

```python
result = EasyCopy.copy_data(
	source=source,
	target=target,
	copy_method="TRUNCATE_APPEND",
	schema_comparison_type="SOFT",
	log_changesets=True,
)
print(result)
```

```python
result = EasyCopy.copy_data(
	source=source,
	target=target,
	copy_method="CHANGEDETECTION",
    id_field="address_id",
	schema_comparison_type="SOFT",
	log_changesets=True,
)
print(result)
```
